### Main concepts of Python DB API
- Connection objects -> database connecitons and managing transactions
- Cursor objects -> send queries to database, scroll through results and retrieving results


#### Examples
- connection methods
  - .cursor, .commit, .rollback, .close()
- cursor methods (fetch operation)
  - execute(), .fetchone() etc.

```python
  
import sqlite3

conn = sqlite3.connect("my_database.db")
cursor = conn.cursor()
```

```python
cursor.execute("DROP TABLE IF EXISTS my_database")
```

```python
#creating a table
table = "create table IF NOT EXISTS my_database(ID INTEGER PRIMARY KEY NOT NULL, FNAME VARCHAR(20), LNAME VARCHAR(20), CITY VARCHAR(20), CCODE CHAR(2))"

cursor.execute(table)

print("Table is Ready")
```

```python
cursor.execute('''insert into my_database values (1, 'Rav', 'Ahuja', 'TORONTO', 'CA')''')
```

```python
cursor.execute('''insert into my_database values (2, 'Raul', 'Chong', 'Markham', 'CA'), (3, 'Hima', 'Vasudevan', 'Chicago', 'US')''')
```

```python
#now we retrieve from the table and data that we've created.

statement = '''SELECT * FROM my_database'''
cursor.execute(statement)

print("All the data")
output_all = cursor.fetchall()
for row_all in output_all:
  print(row_all)
```

```python
#fetching only a few rows

statement = '''SELECT * FROM my_database'''
cursor.execute(statement)
output_many= cursor.fetchmany()
for row_many in output_many:
    print (row_many)
```

```python
statement = '''SELECT FNAME FROM my_database'''
cursor.execute(statement)

print("All the data")
output_column = cursor.fetchall()
for fetch in output_column:
  print(fetch)
```

```python
## updating a column for a person.
query_update = '''update my_database set CITY = 'MOOSETOWN' where FNAME = "Rav"'''
cursor.execute(query_update)

statement = '''SELECT * FROM my_database'''
cursor.execute(statement)
  
print("All the data")
output1 = cursor.fetchmany(1)
for row in output1:
  print(row)
```
